```python
def show_menu(dishes):
    print("=== МЕНЮ ===")
    for index, dish in enumerate(dishes):
        print(f"{index + 1}. {dish}")
    print("=============")

menu = ['Піца', 'Паста', 'Салат']
show_menu(menu)
```

```python
import random as rd

def print_square(n):
    print(n ** 2)

rand_list = set([rd.randint(2,20) for _ in range(10)])
# print(rand_list)

for i in rand_list:
    print_square(i)
```

```python
def find_max(numbers):
    return max(numbers)

numbers = [15, 42, 8, 23, 56, 19]
find_max(numbers)
```
