```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('tip.csv')

df
```

```python
# 1
df['total_bill'].hist(bins=20, edgecolor='black')
plt.title('Total bill distibution')
plt.xlabel('total_bill')
plt.ylabel('quantity');
```

```python
#2
df.boxplot(column='tip', by='day', figsize=(8,5))
plt.title('cost of tickets by class')
plt.suptitle('') # прибрати автоматичний заголовок
plt.ylabel('tip');

# the highest median on sunday
```

```python
# 3
df.plot.scatter(x='total_bill', y='tip', alpha=0.5, c='size', colormap='viridis')
plt.title('Corelation between total bill and tip')
plt.xlabel('total bill')
plt.ylabel('tips');
```

```python
# 4
df.groupby('time')['total_bill'].mean().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Mean bill by day time')
plt.xlabel('Time')
plt.ylabel('Average bill');
```

```python
# 5
import plotly.express as px
fig = px.scatter(df, x='total_bill', y='tip', color='sex',
                 size='size',
                 hover_data=['time', 'day'],
                 title='Scatter plot by total bill and tip')
fig.show()
```

```python
# 6
fig = px.box(df, x='day', y='tip', color='sex',
                 title='Tip distribution by day and sex'
                 )
fig.show()
```
